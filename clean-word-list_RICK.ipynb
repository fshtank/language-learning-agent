```python
# String Processing
import csv
import os

# add Subprocesses that are needed
import subprocess

#String cleaning
from string import punctuation

# Data in a dataform
import pandas as pd

# Data cleanup and adding additional info
import spacy
import spacy_transformers
from wordfreq import zipf_frequency
```

```python
def count_csv_elements_in_file(filepath):

    total_elements = 0

    try:
        with open(filepath, "r", encoding="utf-8") as file:
            csv_reader = csv.reader(file)
            for row in csv_reader:
                total_elements += len(row)
    except (UnicodeDecodeError, IOError):
        # Skip files that can't be read as UTF-8 text
        pass

    return total_elements

language = []
total_words = []

for path, subdirs, files in os.walk("raw-word-lists"):
    for name in files:
        # Only process .txt files; skip .DS_Store and README.md
        if not name.endswith(".txt"):
            continue
        filename = (os.path.join(path, name))
        language += [name.split(".")[0]]
        total_words += [count_csv_elements_in_file(filename)]
```

```python
# Recommended By JunieAI to run Jupyter Notebook

# Provide the path to any target CSV file
sample_file = "raw-word-lists/English/English.txt"  # replace with your target file path

result = count_csv_elements_in_file(sample_file)
print(f"Total elements: {result}")
```

```python
pd.DataFrame({
    "language": language,
    "total_words": total_words
})
```
